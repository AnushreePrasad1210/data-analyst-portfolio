```python
from faker import Faker 
import pandas as pd
import random
fake=Faker()
Faker.seed(42)
random.seed(42)
NUM_RECORDS=50000
product_catalog = [
    "CardioSafe 10mg", "CardioSafe 20mg", "GlucoBalance", "PainRelief Plus",
    "ImmunoBoost", "RespiClear", "NeuroCalm", "DermaHeal", "OrthoFlex", "ThyroCare"
]
sales_records=[]
for transaction_id in range(1,NUM_RECORDS+1):
    product_name=random.choice(product_catalog)
    quantity_sold=random.randint(1,500)
    unit_price=round(random.uniform(50,2000),2)
    total_revenue=round(quantity_sold*unit_price,2)
    transaction_date=fake.date_between(start_date='-2y',end_date='today')
    sales_records.append({
        "transaction_id": transaction_id,
        "transaction_date": transaction_date,
        "product_name": product_name,
        "quantity_sold": quantity_sold,
        "unit_price": unit_price,
        "total_revenue": total_revenue
    })
sales_df = pd.DataFrame(sales_records)
sales_df.head(10)

```

```python
NUM_MRS=50
territory_list = [
    "North Zone", "South Zone", "East Zone", "West Zone",
    "Central Zone", "North-East Zone"
]
mr_records=[]
for mr_id in range(1,NUM_MRS+1):
    mr_name=fake.name()
    assigned_territory=random.choice(territory_list)
    hire_date=fake.date_between(start_date='-8y',end_date='-6M')
    mr_records.append({
        "mr_id": mr_id,
        "mr_name": mr_name,
        "territory": assigned_territory,
        "hire_date": hire_date
    })
mr_df=pd.DataFrame(mr_records)
print(mr_df.head(10))
sales_df["mr_id"]=[random.randint(1,NUM_MRS) for _ in range (len(sales_df))]
sales_df.head(10)
```

```python
NUM_DOCTORS=200
specializations = [
    "Cardiologist", "Endocrinologist", "General Physician",
    "Pulmonologist", "Neurologist", "Dermatologist", "Orthopedic"
]
specialization_weights = [25, 15, 25, 10, 10, 5, 10]
hospital_list = [
    "Apollo Care Hospital", "Fortis Wellness Center", "City General Hospital",
    "Sunrise Multispeciality", "Lakeview Medical Center", "Metro Health Institute"
]
doctor_records=[]
for doctor_id in range(1,NUM_DOCTORS+1):
    doctor_name = "Dr. " + fake.name()
    assigned_specialization=random.choices(specializations,weights=specialization_weights,k=1)[0]
    affiliated_hospital = random.choice(hospital_list)
    doctor_records.append({
        "doctor_id": doctor_id,
        "doctor_name": doctor_name,
        "specialization": assigned_specialization,
        "affiliated_hospital": affiliated_hospital
    })
doctors_df = pd.DataFrame(doctor_records)
print(doctors_df.head(10))
sales_df["doctor_id"] = [random.randint(1, NUM_DOCTORS) for _ in range(len(sales_df))]
sales_df.head(10)

```

```python
messy_sales_df=sales_df.copy()
num_missing_price=int(.02*len(messy_sales_df))
missing_price_indices=random.sample(range(len(messy_sales_df)),num_missing_price)
messy_sales_df.loc[missing_price_indices,"unit_price"]=None
num_missing_qty=int(0.015*len(messy_sales_df))
missing_qty_indices = random.sample(range(len(messy_sales_df)), num_missing_qty)
messy_sales_df.loc[missing_qty_indices, "quantity_sold"] = None
duplicate_indices = random.sample(range(len(messy_sales_df)), 300)
duplicate_rows = messy_sales_df.iloc[duplicate_indices]
messy_sales_df = pd.concat([messy_sales_df, duplicate_rows], ignore_index=True)
num_messy_case = int(0.05 * len(messy_sales_df))
messy_case_indices = random.sample(range(len(messy_sales_df)), num_messy_case)
for idx in messy_case_indices:
    if random.choice([True,False]):
        messy_sales_df.loc[idx, "product_name"] = messy_sales_df.loc[idx, "product_name"].lower()
    else:
        messy_sales_df.loc[idx, "product_name"] = messy_sales_df.loc[idx, "product_name"].upper()
print("Missing values per column:")
print(messy_sales_df.isnull().sum())
print("\nTotal rows now:", len(messy_sales_df))
print("\nSample of inconsistent product names:")
print(messy_sales_df.loc[messy_case_indices, "product_name"].head(5))

```

```python
join_query="""
select
s.transaction_id,
s.transaction_date,
s.product_name,
s.quantity_sold,
s.unit_price,
s.total_revenue,
m.mr_name,
m.territory,
d.doctor_name,
d.specialization
from sales as s
JOIN medical_representatives AS m ON s.mr_id = m.mr_id
JOIN doctors AS d ON s.doctor_id = d.doctor_id
LIMIT 10
"""
joined_df=pd.read_sql(join_query,conn)
print(joined_df)
```

```python
mr_revenue_query = """
SELECT 
    m.mr_id,
    m.mr_name,
    m.territory,
    COUNT(*) AS total_transactions,
    SUM(s.total_revenue) AS total_revenue_generated,
    ROUND(AVG(s.total_revenue), 2) AS avg_revenue_per_sale
FROM sales AS s
JOIN medical_representatives AS m ON s.mr_id = m.mr_id
GROUP BY m.mr_id, m.mr_name, m.territory
ORDER BY total_revenue_generated DESC
"""
mr_revenue_df = pd.read_sql(mr_revenue_query, conn)
print(mr_revenue_df.head(10))
print("\nTotal MRs in result:", len(mr_revenue_df))
```

```python
territory_rank_query = """
SELECT 
    m.mr_id,
    m.mr_name,
    m.territory,
    SUM(s.total_revenue) AS total_revenue_generated,
    RANK() OVER (
        PARTITION BY m.territory 
        ORDER BY SUM(s.total_revenue) DESC
    ) AS rank_within_territory
FROM sales AS s
JOIN medical_representatives AS m ON s.mr_id = m.mr_id
GROUP BY m.mr_id, m.mr_name, m.territory
ORDER BY m.territory, rank_within_territory
"""
territory_rank_df = pd.read_sql(territory_rank_query, conn)
print(territory_rank_df)
```

```python
doctor_trend_query = """
WITH monthly_doctor_revenue AS (
    -- STEP A: Collapse transactions into one row per doctor, per month
    SELECT 
        d.doctor_id,
        d.doctor_name,
        d.specialization,
        strftime('%Y-%m', s.transaction_date) AS sales_month,
        SUM(s.total_revenue) AS monthly_revenue
    FROM sales AS s
    JOIN doctors AS d ON s.doctor_id = d.doctor_id
    GROUP BY d.doctor_id, d.doctor_name, d.specialization, sales_month
)

-- STEP B: Now compare each doctor's month to their OWN previous month
SELECT 
    doctor_id,
    doctor_name,
    specialization,
    sales_month,
    monthly_revenue,
    LAG(monthly_revenue) OVER (
        PARTITION BY doctor_id 
        ORDER BY sales_month
    ) AS previous_month_revenue,
    ROUND(
        (monthly_revenue - LAG(monthly_revenue) OVER (PARTITION BY doctor_id ORDER BY sales_month)) 
        * 100.0 / LAG(monthly_revenue) OVER (PARTITION BY doctor_id ORDER BY sales_month), 
        2
    ) AS pct_change_vs_last_month
FROM monthly_doctor_revenue
ORDER BY doctor_id, sales_month
"""

doctor_trend_df = pd.read_sql(doctor_trend_query, conn)
print(doctor_trend_df.head(20))
at_risk_doctors = doctor_trend_df[doctor_trend_df["pct_change_vs_last_month"] < -50]
print("\nNumber of doctor-months flagged as 'at risk' (revenue dropped >50%):", len(at_risk_doctors))
print(at_risk_doctors.head(10))
```
